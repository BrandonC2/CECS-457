```python
print("tensor x: ")

x= torch.tensor([[1, 3, 0],
                 [2 ,4 , 6]])

print(x)
# -1 means figure this dimension out automatically
print(x.view(-1)) # if you only pass 1 argument to view, it knows you only want a 1 dimensional matrix

print(x.view(3, -1)) # reshape it into three rows

print(x[:, 1:3]) # slicing, index the original array. look at column 1,2
#  : means all
# 1: 3, removes the 0 column
```

CIFAR10 dataset

```python
import torch, torchvision

# run this later
# downloading CIFAR10 dataset 

cifar = torchvision.datasets.CIFAR10('./cifar10/', train = True, download=True)
x = torch.from_numpy(cifar.data).permute(0,3,1,2).float() / 255
print(x.dtype, x.size(), x.min().item(), x.max().item())
```

```python
# narrows to the first images, converts to float
x = x[:48]

# saves these samples as a single image
torchvision.utils.save_image(x, 'cifar-4x12.png', nrow = 12, pad_value = 1.0)
```

```python
print(x.size())
```

```python
# Switches the row and column indexes
x.transpose_(2, 3)
torchvision.utils.save_image(x, 'cifar-4x12-rotated.png',
nrow = 12, pad_value = 1.0)
```

```python
# Kills the green and blue channels
x[:, 1:3].fill_(0)
torchvision.utils.save_image(x, 'cifar-4x12-rotated-and-red.png',
nrow = 12, pad_value = 1.0)
```

```python
# for efficiency reasons, different tensors can share the same data
# and modifying one will modify the other. By default do not make the 
# assumption that two tensors refer to different data in memory

a = torch.full((2, 3), 1)
print(a)

b = a.view(-1)
print(b)

a[1,1] = 2

print(a)
print(b)

b[0] = 9
print(a)
print(b)
```
