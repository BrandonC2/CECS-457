```python
from torch.nn import functional as F
import torch.nn as nn
from matplotlib import pyplot as plt

'''
Please complete the in-class assignment so that

1) You have a single neural network class called Net

2) The constructor initializes all the layers according to the architecture below

3) The forward pass creates a path between the inputs and the outputs

4) Create an instance of Net, and pass a batch of data to it. Verify the output shape. 

For hints on coding this CNN as a PyTorch module, please see the slides ( dlc-slides-4-6-writing-a-module.pdf)
'''

class Net(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(1, 32, kernel_size=5)
        self.conv2 = nn.Conv2d(1, 64, kernel_size=5)
        self.fc1 = nn.Linear(250,200)
        self.fc2 = nn.Linear(200,10)

    def forward(self, x):
        x = F.relu(F.max_pool2d(self.conv1(x), kernel_size=3, stride=3))
        x = F.relu(F.max_pool2d(self.conv2(x), kernel_size=2, stride=2))
        x = x.view(-1, 256)
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x
    



```
