Convolutional NN

The kernel moves over the input vector and calcuates the output.  

for example  
  
input [1, 4, -1, 0, 2, -2, 1, 3, 3, 1]  
  
kernel [1, 2, 0, -1]  
  
the kernel would slide over and calculate the output  

first:  
[1, 4, -1, 0]   
[1, 2, 0, -1]  
==>  
[9, ]

second:  
[4, -1, 0, 2]  
[1, 2, 0, -1]  
==>  
[9, 0]  
  
and repeats
